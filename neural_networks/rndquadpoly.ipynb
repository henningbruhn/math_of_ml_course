# Minima of random quadratic polynomials

We generate random quadratic polynomials in $n$ variables of the form 
$$ p(x_1,\ldots,x_n) = \sum_{i=1}^n\sum_{j=i}^n a_{ij}x_ix_j + \sum_{i=1}^nb_ix_i + c,$$
where all coefficients $a_{ij},b_i,c$ are drawn from independent standard normal distributions.

The Hessian of $p$ is then
$$
H=\begin{pmatrix}
2a_{11} & a_{12} & a_{13} & \ldots & a_{1n}\\
a_{12} & 2a_{22} & a_{23} & \ldots & a_{2n}\\
\vdots &         & \ddots &        & \vdots\\
a_{n1} & a_{n2} & \ldots & \ldots & 2a_{nn}
\end{pmatrix}
$$

A point $x$ is an extremal point if $\nabla p = 0$, which turns out to be equivalent to $Hx=-b$. It turns out that with high probability, $H$ is regular, and there is a unique extremal point $x^*$. That extremal point $x^*$ is a minimum if $H$ is positive definite, ie, if all its eigenvalues are positive.

We do a numerical simulation to estimate the probability that $H$ is positive definite.

```python
import numpy as np

rng = np.random.default_rng(0)
N = 500_000  # samples per n

print(f"{'n':>3}  {'P(local min)':>14}  {'count':>8}")
print("-" * 32)

for n in range(1, 8):
    # Draw all N samples at once: shape (N, n, n)
    A = rng.standard_normal((N, n, n))

    # Build symmetric H using only the lower triangle for independence:
    # off-diagonal entries a_ij ~ N(0,1), diagonal entries 2*a_ii ~ N(0,4)
    H = np.tril(A) + np.swapaxes(np.tril(A, -1), -1, -2)
    H[:, np.arange(n), np.arange(n)] *= 2

    # eigvalsh returns all eigenvalues sorted; check all are positive
    eigvals = np.linalg.eigvalsh(H)       # shape (N, n)
    pos_def = np.all(eigvals > 0, axis=1) # shape (N,)

    p = pos_def.mean()
    print(f"{n:>3}  {p:>14.4f}  {pos_def.sum():>8}")
```
